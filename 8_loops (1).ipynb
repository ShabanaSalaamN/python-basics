# LOOPS IN PYTHON

* A loop is used to repeat a block of code multiple times until a certain condition is met.

* In Python, there are two main types of loops:

    1. `for` loop → used when you know how many times to repeat.

    2. `while` loop → used when repetition depends on a condition.

## Why Loops Are Important

```python
# Without loop
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
```

```python
# using loop
for i in range(5):
    print("Hello")
```

## FOR LOOP

* Used for iterating over a sequence (like a list, string, range, etc.).

* Syntax:
     
           for variable in sequence:
     
                # code to repeat

```python
# range(start, stop, step)
for i in range(5):     # 0 to 4
    print(i)
print("____")

for i in range(1, 6): # 1 to 5
    print(i)
print("____")

for i in range(1, 10, 2):  # Step of 2
    print(i)
print("____")

for i in range(10, 0, -1):  # REVERSE
    print(i)
print("____")

for i in range(5):
    print("HELLO")
print("____")

l=[12,34,56]
for x in range(4):
    print("LIST",l)
```

### Iteration & Performance Tips
- Iterating over characters: `for ch in s:`
- Avoid repeated concatenation in loops; use `join()` for building strings efficiently.

```python
# repeated concatenation
s = ''
for i in range(1000):
    s += 'x'
print(s)
print(len(s))
```

```python
l=[]          #making list using for loop
for i in range(1,10):
    l.append(i)
print(l)
```

```python
# list + join
parts = []
for i in range(1000):
    parts.append('x')
print(parts)
```

```python
parts = []
for i in range(1000):
    parts.append('x')
s2 = ''.join(parts)
print(s2)
print(len(s2))
```

### Nested Loops

```python
# Pattern printing
for i in range(1, 4):
    for j in range(3, 7):
        print(i, j)
```

Rectangle Pattern

```python
for i in range(5):      #the outer loop runs 5 times. Each iteration represents one row.
    for j in range(8):  #For every row, this loop also runs 5 times.
        print("*", end=" ") # using end='', the stars are printed on the same line.
    print()    #prints a blank newline. This moves the cursor to the next line before the next row starts.
```

```python
for i in range(5):      #the outer loop runs 5 times. Each iteration represents one row.
    for j in range(8):  #For every row, this loop also runs 5 times.
        print(j, end=" ") # using end='', the stars are printed on the same line.
    print()    #prints a blank newline. This moves the cursor to the next line before the next row starts.
```

- Outer loop (i) → Number of rows
- Inner loop (j) → Number of columns
- end=" " → Stay on the same line
- print() → Move to the next line

Triangle Pattern

```python
for i in range(1, 6):  #rows
    for j in range(i):  #columns
        print("*", end=" ")
    print()
```

Number Pattern

```python
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

Multiplication Tables

```python
for i in range(1, 6):
    print(f"\nTable of {i}")
    for j in range(1, 11):
        print(f"{i} * {j} = {i*j}")
```

### Loop with else

```python
# Executes when loop finishes normally
for i in range(3):
    print(i)
else:
    print("Loop completed")
```

## WHILE LOOP IN PYTHON

Basic Syntax:

         while condition:
              # code block


* while loop runs as long as the condition is True.

* It stops when the condition becomes False.

```python
# Print numbers from 1 to 5
i = 1
while i <= 5:
    print(i)
    i += 1   # Important: update condition
```

```python
i=2
while i<=20:
    print(i)
    i=i+2     #Print even numbers from 2 to 20
```

```python
i=1
while i<20:
    print(i)
    i+=2        #odd numbers from 1 to 20
```

```python
i=10
while i>=1:
    print(i)
    i-=1        #numbers in reverse (from 10 to 1)
```

```python
i=1
sum=0
while i<=100:
    sum=sum+i
    i=i+1
print("Sum=",sum)      #sum of numbers from 1 to 100
```

```python
# all numbers divisible by 5 between 1 and 100.
i=1
while i<=100:
    if i%5==0:
        print(i,end=" ")
    i=i+1
```

```python
# Print the digits of a number one by one
n=123
while n>0:
    d=n%10
    n=n//10
    print(d)
```

```python
# Count the number of even and odd numbers in a list.
l=[12,34,56,99,31]
even_c=0
odd_c=0
i=0
while i<len(l):
    if l[i]%2==0:
        even_c=even_c+1
    else:
        odd_c=odd_c+1
    i=i+1
print(even_c)
print(odd_c)
```
