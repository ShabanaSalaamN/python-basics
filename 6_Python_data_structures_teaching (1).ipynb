```python
# Create a sample list
data = [10, 20, 30, 20, 40]

print("Original List:", data)

# remove() → Removes first occurrence of value
data.remove(20)
print("After remove (20):", data)

# pop() → Removes element by index (default last)
removed_item = data.pop()
print("Popped item:", removed_item)
print("After pop:", data)

# clear() → Removes all elements
temp = data.copy()   # backup for later use
temp.clear()
print("After clear:", temp)
```

```python
# Create a sample list
data = [10, 20, 30, 20, 40]

print("Original List:", data)

# index() → Returns index of first occurrence
print("Index of 30:", data.index(30))

# count() → Counts occurrences of a value
print("Count of 20:", data.count(20))

# sort() → Sorts the list in ascending order
data.sort()
print("After sort:", data)

# reverse() → Reverses the list
data.reverse()
print("After reverse:", data)

# copy() → Returns a shallow copy
new_list = data.copy()
print("Copied list:", new_list)

# in - check presence of a value
print(20 in new_list)
```

#### List Functions

```python
# Sample list
data = [10, 20, 30, 40, 50]

print("List:", data)

# len() → Number of elements
print("Length:", len(data))

# max() → Largest element
print("Maximum:", max(data))

# min() → Smallest element
print("Minimum:", min(data))

# sum() → Sum of elements
print("Sum:", sum(data))

# sorted() → Returns a new sorted list
print("Sorted list:", sorted(data))

# sorted(reverse=True) → Descending order
print("Descending:", sorted(data, reverse=True))

# any() → True if any element is True
values = [0, 0, 1, 0]
print("Any True?:", any(values))

# all() → True if all elements are True
values2 = [1, 2, 3]
print("All True?:", all(values2))

# list() → Convert to list
t = (1, 2, 3)
converted = list(t)
print("Tuple to list:", converted)
```

## **Tuples**
- Immutable, ordered collection.
- Tuples are faster than lists (read-only data)

```python
# Creating tuple
t = (1, 2, 3)
print(t)
```

```python
# Creating Tuples
t1 = (10, 20, 30)
t2 = 1, 2, 3
t3 = (5,)                # Single element tuple (important!)

print("Tuple 1:", t1)
print("Tuple 2:", t2)
print("Single element tuple:", t3)
```

         * Tuple items can be of any data type
         * To create a tuple with only one item, you have to add a comma after the item,   otherwise Python willnot recognize it as a tuple.

```python
t=(1,2,3,"hello",[1,2],(1,5),"hi",1,True)
print(type(t))
print(t)
```

```python
t=(1)                       # not a tuple, its an int
print(type(t))
tnew=(1,)
print(type(tnew))


t1=("hello")                 # not a tuple , its string
print(type(t1))
t1new=("hello",)
print(type(t1new))


t2=(True)                    # not a tuple, its bool
print(type(t2))
t2new=(True,)
print(type(t2new))


t3=([1,2,3])
print(type(t3))
t3new=([1,2,3],)
print(type(t3new))
```

```python
# It is also possible to use the tuple() constructor to make a tuple.
l=[1,2,3,4,5]
t=tuple(l)              # list will be converted to tuple
print(t)

t1=tuple((1,2,3,4,5,6,7,8,9,"hello"))
print(t1)

t2=tuple({1,4,6,8,"hi"})
print(t2)

t3=tuple(["hello","hi",2+8j,False,1])
print(t3)
```

* Since tuples are immutable we cannot change , add or remove elements from a tuple after creation.

* If we want to do that , first we have to convert the tuple into a list using list() constructor and after that we can chnage , add or remove from the list. After we can convert it back to tuple using tuple() constructor

```python
t=(10,"hello","hey",45,89.5)

#t[0]=12                          # ITS ERROR, NOT POSSIBLE TO DO THIS
                                 # WE CANNOT CHANGE A TUPLE AFTER CREATION
```

```python
t=(10,"hello","hey",45,89.5)
print("Tuple before change:",t)

L=list(t)                              # converted to list
L[0]=12                                # changed the value at position 0
t=tuple(L)                             # Converted back to tuple
print("\ntuple after change:",t)
```

```python
t=(10,"hello","hey",45,89.5)
print("Tuple before adding elements:",t)

L=list(t)                              # converted to list
L.append(12)                           # Added a new value at the end
L.insert(1,"python")                   # inserted a new value at position 1

t=tuple(L)                             # Converted back to tuple
print("\ntuple after adding elements:",t)
```

```python
t=(10,"hello","hey",45,89.5,12)
print("Tuple before removing elements:",t)

L=list(t)                              # converted to list
L.remove("hey")                        # removes the element hey
L.pop(1)                               # deleting an element from the position 1

t=tuple(L)                             # Converted back to tuple
print("\ntuple after remoivng elements:",t)
```

## **Sets**
- A set is an unordered, mutable collection of unique elements(no duplicates).

```python
# Using curly braces
s1 = {1, 2, 3, 4}

# Using set() function
s2 = set([1, 2, 3, 4])

# Empty set (important!)
empty = set()     # {} creates a dictionary, not a set

print(s1, s2, empty)
```

```python
s={12,13.5,"hello",(2,"hi"),True,False,2+3j,None}
print(s)
```

#### Properties of Sets

```python
s = {10, 20, 20, 30}

# Unordered → no indexing
print(s)

# No duplicates → automatically removed
print("Duplicates removed:", s)

# Mutable → can add/remove elements
```

```python
# Creating set
s = {1, 2, 3, 3}
print(s)  # duplicates removed
```

* False and 0 will be considered as duplicates : so if you give both in a set, the first one will be taken

#### Set operations

```python
A = {1, 2, 3}
B = {3, 4, 5}

# Union
print("Union:", A | B)

# Intersection
print("Intersection:", A & B)

# Difference
print("A - B:", A - B)

# Symmetric Difference
print("Symmetric Difference:", A ^ B)
```

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
print(A.symmetric_difference(B))
```

```python
# union() method:  Joins items from two or more sets
s1={10,20,30,40,"hello"}
s2={50,60,"hi","hello",40}
s3=s1.union(s2)                               # joined s2 with s1 into a new set s3
print(s3)                                     # duplicates will be neglected
```

```python
# Joining multiple sets
s1={1,2,3,4}
s2={3,4,5,6,7}
s3={7,8,9,10}
s4={10,11,9,12}
s5=s1.union(s2,s3,s4)

print(s5)
```

```python
# We can use | operator instead of union() methods
s1={1,2,3,4}
s2={3,4,5,6,7}
s3={7,8,9,10}
s4={10,11,9,12}
s5=s1|s2|s3|s4
print(s5)
```

# Mini Project: Student Record Management System

Objective

Build a simple system to:

- Add student details
- View all students
- Search student
- Update marks
- Delete student

```python
# Use a dictionary of dictionaries

# Student Record Management System

students = {}

# 🔹 Add student
def add_student():
    sid = input("Enter Student ID: ")
    name = input("Enter Name: ")
    marks = int(input("Enter Marks: "))

    students[sid] = {"name": name, "marks": marks}
    print("Student added successfully!\n")


# 🔹 View all students
def view_students():
    if not students:
        print("No records found\n")
        return

    for sid, details in students.items():
        print(f"{sid} -> Name: {details['name']}, Marks: {details['marks']}")
    print()


# 🔹 Search student
def search_student():
    sid = input("Enter Student ID to search: ")

    if sid in students:
        print(students[sid])
    else:
        print("Student not found\n")


# 🔹 Update marks
def update_marks():
    sid = input("Enter Student ID: ")

    if sid in students:
        new_marks = int(input("Enter new marks: "))
        students[sid]["marks"] = new_marks
        print("Marks updated\n")
    else:
        print("Student not found\n")


# 🔹 Delete student
def delete_student():
    sid = input("Enter Student ID: ")

    if sid in students:
        students.pop(sid)
        print("Student deleted\n")
    else:
        print("Student not found\n")


# 🔹 Menu-driven program
while True:
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update Marks")
    print("5. Delete Student")
    print("6. Exit")

    choice = input("Enter choice: ")

    if choice == '1':
        add_student()
    elif choice == '2':
        view_students()
    elif choice == '3':
        search_student()
    elif choice == '4':
        update_marks()
    elif choice == '5':
        delete_student()
    elif choice == '6':
        print("Exiting program...")
        break
    else:
        print("Invalid choice\n")
```
