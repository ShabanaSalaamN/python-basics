# Control Statements in Loops:
`break, continue, pass`


- `break` → exit loop
- `continue` → skip iteration
- `pass` → placeholder

## `break`


Stops the loop immediately.

```python
# Task: Stop loop when number becomes 5
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

```python
# Search Operation

numbers = [10, 20, 30, 40, 50]
search = 30

for n in numbers:
    if n == search:
        print("Number Found")
        break
```

```python
# Keep asking the user for input until they enter "stop".
while True:
    user=input("Enter")
    if user=="stop":
        break
```

```python
# Take input until user enters 0
while True:
    num = int(input("Enter number (0 to stop): "))
    if num == 0:
        break
    print("You entered:", num)
```

```python
# Stop on Negative Number
while True:
    num = int(input("Enter number: "))
    if num < 0:
        print("Negative number entered")
        break

    print("You entered:", num)
```

```python
# Ask the user to enter numbers repeatedly.
# Count how many positive numbers they entered.
# When user enters -1, stop and print the count.
count=0
while True:
    n=int(input("enter numbers"))
    if n>0:
        count=count+1
    if n==-1:
        print(count)
        break
```

```python
# Keep asking the user for numbers and add them to a total.
# When the user enters "done", stop and print the total sum.
total=0
while True:
    n=int(input("Enter number"))
    total=total+n
    user=input("Enter done or not")
    if user=="done":
        print(total)
        break
```

```python
# ATM Program
balance = 10000
while True: #Since True is always true, the condition never becomes false.Therefore, the loop keeps executing.
    amount = int(input("Enter withdrawal amount: "))
    if amount <= balance:
        balance -= amount
        print("Transaction Successful")
        print("Remaining Balance =", balance)
        break
    else:
        print("Insufficient Balance")
```

```python
# Password Validation
correct_password = "python123"

while True:
    password = input("Enter password: ")
    if password == correct_password:
        print("Login Successful")
        break
    else:
        print("Wrong Password")
```

```python
# To validate your username and password
user="aswathi123"
pasw=12345
while True:
    username=input("Enter your username")
    password=int(input("Enter your password"))
    if username==user and password==pasw:
        print("LOGIN SUCCESSFULL")
        break
    else:
        print("ENTER CORRECT PASSWORD OR USERNAME")
```

```python
# Guessing Game
secret = 7

while True:
    guess = int(input("Guess the number: "))

    if guess == secret:
        print("Correct Guess")
        break

    print("Try Again")
```

```python
# Break in Nested Loop
for i in range(1, 4):
    for j in range(1, 6):
        if j == 3:
            break
        print(j)
    print("Outer Loop")
```

## `continue`

Skips current iteration.


```python
# Task: Skip number 3
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
# Modify the code: Skip even numbers / Print Odd Numbers Only
for i in range(1, 21):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# Skip Negative Numbers
numbers = [10, -5, 20, -8, 30]
for n in numbers:
    if n < 0:
        continue
    print(n)
```

```python
# Skip Multiples of 3
for i in range(1, 31):
    if i % 3 == 0:
        continue
    print(i)
```

```python
# Student Attendance Example
students = ["Rahul", "Anu", "Arjun", "Diya"]
absent = "Arjun"

for student in students:
    if student == absent:
        continue
    print(f"{student} is present")
```

```python
# Skip Failed Students
marks = [78, 35, 90, 25, 88]
for m in marks:
    if m < 40:
        continue
    print(f"{m} -> Pass")
```

```python
# Quiz Program
score = 0
while True:
    answer = input("Capital of India: ")
    if answer == "":
        continue
    if answer.lower() == "delhi":
        score += 1
        print("Correct")
        break
    else:
        print("Wrong Answer")
print("Score =", score)
```

## `pass`
Placeholder statement.

pass means:
“Do nothing for now.”

```python
# Task: Run empty loop
for i in range(5):
    pass
print("Loop completed")
```

```python
# Empty Function
def my_function():
    pass
```

```python
# Empty If Block
age = 20
if age >= 18:
    pass
else:
    print("Minor")
```

## Combined Example
Observe break + continue together

```python
for i in range(1, 10):
    if i == 3:
        continue
    if i == 7:
        break
    print(i)
```

```python
while True:
    num = int(input("Enter number (0 to stop): "))
    if num == 0:
        break
    if num < 0:
        continue
    print("Valid number:", num)
```

```python
# Continuously ask the user to enter a number.
# If the number is negative → skip it using continue.
# If the number is 0 → exit using break.
# Otherwise → print its square.
while True:
    n=int(input("Enter a number"))
    if n<0:
        continue
    elif n==0:
        break
    else:
        print(n*n)
```

```python
# Student Marks Average
total = 0

n = int(input("How many subjects? "))

for i in range(n):

    marks = float(input(f"Enter marks for subject {i+1}: "))

    total += marks

average = total / n

print(f"Average Marks = {average:.2f}")
```

```python
# Electricity Bill Generator
total = 0
days = int(input("Enter number of days: "))

for i in range(days):
    units = int(input(f"Units consumed on day {i+1}: "))
    total += units
bill = total * 8

print(f"Total Units = {total}")
print(f"Electricity Bill = ₹{bill}")
```

```python
# Prime Number Check
num = int(input("Enter number: "))
is_prime = True

for i in range(2, num):
    if num % i == 0:
        is_prime = False
        break
if is_prime:
    print("Prime Number")
else:
    print("Not Prime")
```

```python
# Fibonacci Series
n = int(input("How many terms? "))
a = 0
b = 1

print(a)
print(b)

for i in range(2, n):
    c = a + b
    print(c)
    a = b
    b = c
```

```python
# Guess the Number Game
secret = 7

while True:
    guess = int(input("Guess the number: "))
    if guess == secret:
        print("Correct Guess")
        break
    else:
        print("Try Again")
```
