#Slicing and indexing

```python
l= ["hello","hi",23,45.6,True,12,78,False,[1,2,3,4,5]]

print(l[3])           
print(l[1])

print(l[-1])
print(l[-2])

print(l[2:4])
print(l[-5:-1])

print(l[-1:-6:-1])
print(l[::-1])
print(l[-1::-1])
print(l[:-1])
print(l[-1][0:3])
```

#Nested Lists

```python
matrix=[[1,2,3],[4,5,6],[7,8,9]]
print(matrix)
```

```python
matrix[1][2]
```

```python
matrix[2][0:2]
```

#extend to extend a list with the element of another list

```python
l1= [1,2,3,4,"hello"]
l2= [5,6,7,8,9]
t= (10,11,12,13)
l1.extend(l2)
print(l1)
```

```python
l1.extend(t)
print(l1)
```
