## **f-Strings**

```python
name = "Anita"
marks = 92

print(f"{name} scored {marks} marks.")
```

- f stands for formatted string
- Variables can be directly used inside {}

### Expressions Inside f-Strings

```python
a = 10
b = 20

print(f"Sum = {a + b}")
print(f"Multiplication = {a * b}")
```

### Formatting Decimal Values

```python
pi = 3.1415926535

print(f"Value of pi = {pi:.2f}")
# .2f means 2 digits after decimal
```

### Aligning Text

```python
name = "Rahul"
score = 88

print(f"{name:<10} {score}")
print(f"{name:>10} {score}")
print(f"{name:^10} {score}")
```

- `<` Left align
- `>` Right align
- `^` Center align

### Percentage Formatting

```python
marks = 87.4567

print(f"Percentage = {marks:.1f}%")
```

```python
x = int(input("Enter first number: "))
y = int(input("Enter second number: "))

print("Before swapping: x =", x, "y =", y)

x, y= y,x

print("After swapping: x =", x, "y =", y)
```

```python
name = "Alice"
marks = 85

# Method 1: Using comma
print("Name:", name, "Marks:", marks)

# Method 2: f-string (best)
print(f"Name: {name}, Marks: {marks}")

# Method 3: format()
print("Name: {}, Marks: {}".format(name, marks))
```

```python
PI = 3.14159
radius = float(input('Enter radius: '))
area = PI * radius ** 2
print('Area:', area)
```

```python
name = input("Enter your name: ")
python_marks = float(input("Enter Python marks: "))
sql_marks = float(input("Enter SQL marks: "))

average = (python_marks + sql_marks) / 2

print("\n===== REPORT CARD =====")

print(f"Student Name : {name}")
print(f"Python Marks : {python_marks:.2f}")
print(f"SQL Marks    : {sql_marks:.2f}")
print(f"Average      : {average:.2f}")
```

```python
product = input("Enter product name: ")
price = float(input("Enter product price: "))
quantity = int(input("Enter quantity: "))

total = price * quantity

print("\n========== BILL ==========")
print(f"Product Name : {product}")
print(f"Price        : ₹{price:.2f}")
print(f"Quantity     : {quantity}")
print(f"Total Amount : ₹{total:.2f}")
print("==========================")
```
