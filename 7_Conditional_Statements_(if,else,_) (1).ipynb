# **Conditional statements**

Conditional statements help programs make decisions.

In Python, we mainly use:

- if
- if-else
- if-elif-else
- Nested if
- Multiple conditions using and, or, not

## **if**


* You can write an if statement in one line if it has only one action.



```
if condition:
    statement
```



````python
age = 20

if age >= 18:
    print("You are eligible to vote")
````

- Condition checked: age >= 18
- If condition is True, block executes

````python
num = int(input("Enter a number: "))

if num > 0:
    print("Positive Number")
````

````python
x=10
if x > 5:
    print("x is greater than 5")
````

````python
x=10
if x>5: print("x is greater than 5")
````

Executes only if condition is True

## **if - elif - else**

Used when there are multiple conditions.



```
if condition1:
    statements

elif condition2:
    statements

elif condition3:
    statements

else:
    statements
```



````python
a,b=10,20
if a==b:
    print("Equal")
elif a>b:
    print("a is bigger")
else:
    print("b is bigger")
````

````python
'equal' if a==b else 'a is bigger' if a>b else 'b is bigger'
````

````python
# Largest of Three Numbers

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a > b and a > c:
    print("A is largest")

elif b > a and b > c:
    print("B is largest")

else:
    print("C is largest")
````

## **Nested if**

An if inside another if


```
if condition1:

    if condition2:
        statements
```



````python
# Check positive AND even

num = 8

if num > 0:
    print("Positive number")

    if num % 2 == 0:
        print("Even number")
    else:
        print("Odd number")

else:
    print("Negative number")
````

6. Input three angles →

    * If sum = 180 → "Valid triangle"

        * If all angles equal → "Equilateral"

        * Else if any two equal → "Isosceles"

        * Else → "Scalene"

    * Else → "Invalid triangle".

````python
angle1=float(input("Enter angle1"))
angle2=float(input("Enter angle2"))
angle3=float(input("Enter angle3"))

if angle1+angle2+angle3==180:
    print("VALID TRIANGLE")
    if angle1==angle2==angle3:
        print("TRIANGLE IS EQUILATERAL")
    elif angle1==angle2 or angle1==angle3 or angle2==angle3:
        print("TRIANGLE IS ISOSCELES")
    else:
        print("TRIANGLE IS SCALENE")
else:
    print("NOT A VALID TRAINGLE")
````

5. Input a year and check whether it is a leap year.

   * A leap year is a year that contains 366 days instead of the usual 365, with an extra day added to February (February 29).

   * A year is a leap year if:
   
        * It is divisible by 4 AND
             
          * Not divisible by 100, unless it is also divisible by 400.

````python
y=int(input("Enter a year"))
if (y%4==0 and y%100!=0) or y%400==0:
    print(f"{y} is a leap year")
else:
    print(f"{y} is not a leap year")
````

BMI Calculator

````python
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in meters: "))

bmi = weight / (height ** 2)

print(f"BMI = {bmi:.2f}")

if bmi < 18.5:
    print("Underweight")

elif bmi < 25:
    print("Normal Weight")

elif bmi < 30:
    print("Overweight")

else:
    print("Obese")
````

Student Result System

````python
name = input("Enter student name: ")

python_marks = int(input("Enter Python marks: "))
sql_marks = int(input("Enter SQL marks: "))
ml_marks = int(input("Enter ML marks: "))

total = python_marks + sql_marks + ml_marks
average = total / 3

print("\n===== RESULT =====")

print(f"Student Name : {name}")
print(f"Total Marks  : {total}")
print(f"Average      : {average:.2f}")

if average >= 90:
    print("Grade A")

elif average >= 75:
    print("Grade B")

elif average >= 50:
    print("Grade C")

else:
    print("Fail")
````
